```python
from math import isqrt
def snt(a):
    for i in range(2,isqrt(a)+1):
        if (a%i==0 and a!=2):
            return False
    if a==0 or a==1:
        return False
    return True
n=int(input())
while n<0:
    print(n)
    n=int(input())
if snt(n):
    print('Yes')
else:
    print('No')
```
