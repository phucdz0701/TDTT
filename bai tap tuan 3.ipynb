Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình

```python
from math import *
a,b,c=map(float,input().split())
p=(a+b+c)/2
if a+b<=c or a+c<=b or b+c<=a:
    print('Khong la 3 canh cua mot tam giac')
else:
    if a==b==c:
        print('Day la tam giac deu')
    elif a==b!=c or a==c!=b or b==c!=a:
        print('Day la tam giac can')
    elif a**2+b**2==c**2 or a**2 + c**2 == b**2 or a**2==b**2 + c**2:
        print('day la tam giac vuong')
    print('dien tich tam giac la:',sqrt(p*(p-a)*(p-b)*(p-c)))
    print('chu vi tam giac la:',p*2)
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
from math import *
a,b,c=map(float,input('nhap a,b,c').split())
delta=b**2-4*a*c
if delta==0:
    print('x=',-b/(2*a))
elif delta<0:
    print('Phuong trinh vo nghiem')
else:
    print('x1=',(-b+sqrt(delta))/(2*a))
    print('x2=',(-b-sqrt(delta))/(2*a))
```

Bài 8: Giải hệ phương trình tuyến tính:
ax+by=m
cx+dy=n

```python
a,b,m,c,d,n=map(float,input().split())
D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c
if D!=0:
    print('x=',f'{Dx/D:.2f}')
    print('y=',f'{Dy/D:.2f}')
elif D==0 and (Dx!=0 or Dy!=0) (a == b == 0 and m != 0) or (c == d == 0 and n != 0):
    print('Vo nghiem')
else:
    print('Vo so nghiem')
```
